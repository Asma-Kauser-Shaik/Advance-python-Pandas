```python
# add new columms in dataset
# rank function provide the rank based upon the salary
# lower salary = lower rank
df['rank']=df.salary.rank()
df
```

```python
# sorting 
df['rank']=df.salary.rank(ascending = False)
df.sort_values('rank')
# if column already present then this will update the column
# if column don't exsist then new column will be created
```

```python
# update column already present column
df['rank123']=df.salary.rank(ascending = False)
df.sort_values('rank123')
```
