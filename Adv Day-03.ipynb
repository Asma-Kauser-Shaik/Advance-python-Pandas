# Data Filtering and indexing

```python
# Query: find those records where salary morethan 1 lakh
df.salary > 100000
```

```python
# to find the records and total no of person.
df[df.salary > 100000]
```

```python
df[df.salary<50000] # to find the salary who get less than 50000
```

```python
df[(df.salary>50000)& (df.salary<100000)]
```

```python
# shape is used to print only data or count 
df[(df.salary>50000)& (df.salary<100000)].shape
```

```python
# query : 
df[(df.salary< 50000)|(df.salary>150000)]
```

```python
df[(df.salary< 50000)|(df.salary>150000)].shape
```

```python
# query: to find the max salary  person from the records.
df[df.salary == df.salary.max()]
```

```python
# query: to find the min salary person from the record.
df[df.salary == df.salary.min()]
```

```python
# query : find the record of the person who got max marriage propsals.
df[df.Marriage_proposal_count == df.Marriage_proposal_count.max()]
```

```python
# we need only specific data .
df[df.Marriage_proposal_count == df.Marriage_proposal_count.max()][['Name','Profession']]
```

# Indexing

```python
# loc & iloc
```

```python
# Query: print 1 to 6 records and Name to profession columns
```

```python
df.columns
```

```python
# df.loc[row,columns] 
# loc is used for filtering the fixed no of columns and rows.
df.loc[1:6,'Name':'Profession']
```

```python
# Query: print 10 to 20 records and Age to salary columns
df.loc[10:20,'Age':'salary']
```

```python
# when we want print two spefic
df.loc[10:20,['Age','salary']]
```

```python
# we can use step too
df.loc[10:20:2,['Age','salary']]
```

```python
# iloc 
# Query: print record from 1 to 6 and columns from 1 to 3
df.iloc[1:7,1:4]
```

```python
df.iloc[0:11,0:6]
```

```python
# different blw loc and iloc 
# in loc we need  to mention the column names
# while in iloc we mention the index of columns
# iloc goes upto n-1 ,but loc goes upto nth value
# iloc is faster than loc 
# because loc need to convert column names to index.
```

```python
df.iloc[0,1]
```

```python
df.iloc[0,2]
```
