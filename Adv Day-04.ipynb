# Handling Null values

```python
import pandas as pd
```

```python
df = pd.read_excel("excel_data.xlsx")
df
```

```python
# Null : the information which is not provided
# we define the null value in pandas as (NaN,na)
# Null vs zero 
# null is nothing.
# zero is something value.
# null is no value ,while null means no value avaliable.
```

```python
# identify the null values in dataframe.
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
# find the null values in a specific column
df.Age.isnull().sum()
```

```python
# print the records with null values
df.Age.isnull()
```

```python
df[df.Age.isnull()]
```

```python
# delete the records which is having null values.
df1 = df.copy()
df1.dropna(inplace = True)
```

```python
df1.isnull().sum()
```

```python
df1.shape
```

```python
# approach-1: delete the records which is having null values.
# approach -2 : delete the column which is having the null values.
df2 = df.copy()
```

```python
df2.isnull().sum()
```

```python
df2 = df2.dropna(axis = 1)# to delete column we use axis = 1 & we want to delete row we use axis = 0
df2.isnull().sum()
```

```python
# how = any :delete the column even that have single null values
# how = all : delete the column only if all the values in that column is Null
df2 = df2.dropna(axis = 1,how = 'any')
df2.isnull().sum()
```

```python
df2 = df2.dropna(axis = 1,how = 'all')# but my output delete age but age is not delete/.
df2.isnull().sum()
```

```python
# Approach-3:
# replace the null value with some constant
df3 = df.copy()
```

```python
# fillna() function is used to fill the null values with some other value
df3.Age = df3.Age.fillna(18)
df3.isnull().sum()
```

```python
df3.shape
```

```python
# filling age value with average value
avg_age = df.Age.mean()
df.Age = df.Age.fillna(avg_age)
df.Age.isnull().sum()
```

# handling the Duplicate values

```python
# identify the duplicate records
df.duplicated()
```

```python
# to find sum of duplicate function
df.duplicated().sum()
```

```python
# show the duplicate records
# if a record is completely matching with some other record
df[df.duplicated()]
```

```python
# by this we see the all matching records
# if keep value is false : show the both records.
# first : marks duplicates as 'true' except for the first occurrence.
# last:marks duplicates as 'true' except for the last occurrence.
df[df.duplicated(keep = False)]
```

```python
df[df.duplicated(keep ='first')]
```

```python
df[df.duplicated(keep ='last')]
```

```python
# deleting the duplicate records
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```
