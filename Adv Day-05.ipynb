# Data transformation using apply()& map()function

```python
# apply(): this function is used to execute lambda function 
# lambda function is one line function 
# if you want to apply lambda function on dataframe then we use the apply() function 
```

```python
df1 = df.copy()
```

```python
# on the festival occation : sai ram decided to double the empolyee salary
df1.salary = df1.salary.apply(lambda x : x*2)
df1
```

```python
df.columns = df.columns.str.title()
```

```python
df.columns
```

```python
# create a new column Bonus or Non- Bonus
# Apply() function with complete dataframe
df1["Diwali_gift"] = df1.apply(lambda row : 'Bonus' if row['salary'] < 100000 else 'No_Bonus' ,axis = 1)
df1
```

```python
df1["Diwali_gift"] = df1.apply(lambda row : 'Bonus' if (row['salary'] and row['age']> 40) < 20000 else 'No_Bonus' ,axis = 1)
df1
```
