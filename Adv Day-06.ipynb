## merging of Dataframe

```python
# merge() : this function can be used to merge the dataframes based upon common column
# merge()function work like join in sql : Inner,outer,left,Right
df1 = pd.DataFrame({'ID' : [1,2,3],'Dream_company':['Google','Microsoft','Apple']})
df2 = pd.DataFrame({'ID':[2,3,4], 'Dream_package':['8lac','10lac','12lac']})
merge_df = pd.merge(df1 , df2, on = 'ID',how = 'inner')# inner join
merge_df
```

```python
merge_df = pd.merge(df1 , df2, on = 'ID',how = 'outer')# outer join
merge_df
```

```python
merge_df = pd.merge(df1 , df2, on = 'ID',how = 'right')# right join
merge_df
```

```python
merge_df = pd.merge(df1 , df2, on = 'ID',how = 'left')# left join
merge_df
```

```python
df1 = pd.DataFrame({'ID' : [1,2,3],'Dream_company':['Google','Microsoft','Apple'],'Name':['Asma','kauser','abdul']})
df2 = pd.DataFrame({'ID':[2,3,4], 'Dream_package':['8lac','10lac','12lac'],'Name':['Asma','shan','kauser']})
merge_df = pd.merge(df1 , df2, on = 'Name',how = 'inner')# inner join
merge_df
```

```python
merge_df = pd.merge(df1 , df2, on = 'Name',how = 'outer')
merge_df
```

```python
merge_df = pd.merge(df1 , df2, on = 'Name',how = 'left')
merge_df
```

```python
merge_df = pd.merge(df1 , df2, on = 'Name',how = 'outer')# outer join
merge_df
```

```python
# join(): mainly used to work with index.
df3 = pd.DataFrame({"Hero_Name" : ["NTR","MB","AA"]},index = [1,2,3])
df4 = pd.DataFrame({"Herione_Name" : ["sam","Rakul","pooja"]},index = [2,3,4])
joined_df = df3.join(df4, how = 'inner')
joined_df
```

```python
joined_df = df3.join(df4, how = 'outer')
joined_df
```

```python
joined_df = df3.join(df4, how = 'inner')
joined_df
```

```python
joined_df = df3.join(df4, how = 'right')
joined_df
```

```python
joined_df = df3.join(df4, how = 'left')
joined_df
```

## concat()

```python
# def : concat() function is used to stack dataframes vertically or horizontally
```

```python
df5 = pd.DataFrame({'ID':[1,2],'Fav_moive':["Paraside","Hit-3"]})
df6 = pd.DataFrame({'ID':[3,4],'Fav_moive': ["Hello","Spiderman"]})
# axis = 0 : vertically (adding the rows)
# axis = 1 : vertically (adding the columns)
concat_df = pd.concat([df5,df6],axis = 0)
concat_df
                    
```

```python
concat_df = pd.concat([df5,df6],axis = 1)
concat_df
         
```

```python
concat_df = pd.concat([df5,df6],axis = 0,ignore_index = True)# to reset the index numbers after combing the dataframe.
concat_df
             
```
