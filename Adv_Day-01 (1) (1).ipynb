# series data strture

```python
# Series is a 1D Homogeneous array.
```

```python
ser1 = pd.Series() # empty series
print(ser1)
print(type(ser1))
```

```python
ser2 = pd.Series([10,20,30,40,50])
print(ser2)
print(type(ser2))
```

```python
ser2 = pd.Series([10,20,30,40,50,"Hello"])
print(ser2)
print(type(ser2))
# String in pandas are represented as "object".
# Internally,pandas convert all the elements to string if even a single element passed is string
```

```python
ser3 = pd.Series(["Apple","Jack_fruit","Mango"])
print(ser3)
```

```python
# pandas series don't support the negative  indexing.
ser3[-1]
```

# Dataframe: 2D array

```python
df1 = pd.DataFrame() # empty array
print(df1)
```

```python
# one columns data frame 
df2 = pd.DataFrame(["Asma","Kauser","Abdul"],columns = ["Name"])
df2
```

```python
# two columns data frame
df2 = pd.DataFrame([["Asma",100],["Kauser",200],["Abdul",300]],columns = ["Name","Num"])
df2
```

```python
import pandas as pd

df2 = pd.DataFrame(
    [
        ("Asma", 22, 5.6),
        ("Kauser", 23, 6.0),
        ("Abdul", 25, 6.2)
    ],
    columns=("Name", "Age", "Height")
)

df2
```

```python
# convert list into tuple and 3 columns
df2 = pd.DataFrame([("Asma",22,5.6),("Kauser",23,6.0),("Abdul",25,6.2)],columns = ("Name","Age","Height"))
df2
```

```python
# convert the dictionary into dataframe.
import pandas as pd
fav_hero ={
    "Hero_Name" : ["gopichand","Pk","MB","AA"],
    "Salary": [30,40,50,60]
}
df4 = pd.DataFrame(fav_hero)
df4
```
